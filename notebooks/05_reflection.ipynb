# V. Tổng Kết và Phản Ánh

Phần này tóm tắt những phát hiện chính, hạn chế của dự án, hướng phát triển trong tương lai, và những trải nghiệm cá nhân của từng thành viên.


---

## 5.1 Tóm Tắt Dự Án

### A. Các Phát Hiện Chính

**1. Nội dung tường minh không làm giảm độ phổ biến**
- Các bài hát có nội dung tường minh đạt mức độ phổ biến cao hơn đáng kể (57.5 vs 50.5)
- Nghệ sĩ không cần tự kiểm duyệt để đạt được thành công thương mại

**2. Khoảng thời lượng vàng: 3.5-4.5 phút**
- Bài hát trong khoảng này đạt popularity cao nhất (~55.6-55.9)
- Bài hát quá ngắn (<2p) hoặc quá dài (>5p) có độ phổ biến thấp hơn

**3. Album vượt trội hơn single**
- Full album có độ phổ biến cao nhất (55.5), vượt single (46.1) và compilation (40.5)
- Người nghe có xu hướng ưa thích trải nghiệm album trọn vẹn

**4. Số lượng bài hát ảnh hưởng đến độ nổi tiếng nghệ sĩ**
- Nghệ sĩ phát hành nhiều (>100 tracks) đạt popularity gần gấp đôi nghệ sĩ ít (1-20 tracks)
- Việc phát hành đều đặn quan trọng hơn hoàn thiện một vài bài

**5. Yếu tố thành công khác nhau theo từng phân khúc**
- Phân khúc thấp: Danh tiếng nghệ sĩ là yếu tố chính ($R^2$=0.47)
- Phân khúc cao: Khó dự đoán, phụ thuộc nhiều vào may mắn và viral ($R^2$=0.03)

**Phát hiện thú vị nhất:** Nhạc cũ (2009-2015) vẫn duy trì độ phổ biến tương đương nhạc mới (2020-2025), chứng tỏ giá trị của catalog âm nhạc là lâu dài.


### B. Hạn Chế

**Về dữ liệu:**
- Chỉ có 8,772 tracks, không đại diện cho toàn bộ hệ sinh thái âm nhạc
- Thiếu thông tin quan trọng: marketing budget, playlist placement, mạng xã hội, âm thanh học
- Bị thiên lệch về các bài hát phổ biến và nghệ sĩ đã thành danh

**Về phân tích:**
- Chỉ tìm được tương quan, không chứng minh nhân quả
- Mô hình tốt nhất chỉ đạt $R^2$=0.32, còn 68% phương sai chưa giải thích
- Không phân tích sâu theo thể loại âm nhạc riêng biệt

**Về phạm vi:**
- Không xem xét yếu tố địa lý và văn hóa
- Thiếu phân tích theo thời gian (chỉ chụp ngang tại một thời điểm)
- Không có thông tin về hành vi người dùng và tâm lý nghe nhạc

### B. Phân tích sâu hơn

- Sử dụng phương pháp suy luận nhân quả để tách biệt các hiệu ứng
- Áp dụng học sâu cho dự đoán chính xác hơn
- Phân tích âm thanh học từ file âm thanh thực tế

### C. Dữ liệu bổ sung cần thu thập

- Số liệu streaming thực tế, tỷ lệ bỏ qua, tỷ lệ lưu bài
- Dữ liệu marketing và quảng cáo
- Dữ liệu mạng xã hội (TikTok, Instagram, YouTube)
- Các đặc trưng âm thanh học chi tiết
- Nội dung lời bài hát và phân tích cảm xúc

### D. Mở rộng và cải tiến

- Tăng quy mô dataset lên hàng triệu bài hát
- So sánh giữa các nền tảng streaming khác nhau

## 5.3 Phản Ánh Cá Nhân

### A. Phạm Phú Hòa


#### Những khó khăn gặp phải

**Khó khăn kỹ thuật:** Khó khăn trong việc chọn đặc trưng hợp lý từ dữ liệu có sẵn. Dataset thiếu nhiều thông tin quan trọng (đặc trưng âm thanh, marketing, vị trí trong danh sách phát) nên $R^2$ chỉ đạt ~0.36.

**Khó khăn phân tích:**
- **Tìm độ dài vàng**: Phải thử nghiệm nhiều cách phân khoảng khác nhau (1 phút, 0.5 phút, 0.25 phút) để tìm khoảng thời lượng vàng chính xác. Cân nhắc giữa độ chi tiết và kích thước mẫu.
- **Giải thích độ quan trọng đặc trưng**: Khó giải thích tại sao `artist_popularity` quan trọng hơn `track_duration_min`, phải phân tích sâu hơn về mối quan hệ giữa các biến.
- **Lựa chọn mô hình**: So sánh nhiều mô hình (Hồi quy tuyến tính, Random Forest, Gradient Boosting, Extra Trees) để tìm mô hình tốt nhất cho từng phân khúc độ phổ biến.

**Giải quyết:**
- Tham khảo ví dụ về độ quan trọng đặc trưng và cách giải thích mô hình
- Làm việc nhóm để brainstorm các đặc trưng có thể tạo ra từ dữ liệu hiện có

#### Học hỏi và phát triển

**Kỹ năng kỹ thuật:**
- **Pandas & NumPy**: Thành thạo hơn trong xử lý dữ liệu, nhóm dữ liệu, phân khoảng, gộp tập dữ liệu
- **Scikit-learn**: Hiểu sâu về quy trình máy học, chia dữ liệu, xác thực chéo, chuẩn hóa, các mô hình hồi quy
- **Git & cộng tác**: Làm quen với commit, push, xử lý xung đột, quy trình làm việc nhóm

**Hiểu biết về Máy học:**
- **Xác thực chéo**: Hiểu tầm quan trọng của xác thực chéo K-Fold trong việc đánh giá độ tin cậy mô hình, tránh quá khớp
- **Rò rỉ dữ liệu**: Học được khái niệm nghiêm trọng của rò rỉ dữ liệu và cách phòng tránh
- **Độ quan trọng đặc trưng**: Hiểu cách các mô hình cây quyết định tính toán độ quan trọng đặc trưng và ý nghĩa thực tế
- **Giải thích mô hình**: $R^2$ cao không phải lúc nào cũng tốt, phải xem xét bối cảnh và các đặc trưng còn thiếu

### B. Nguyễn Lâm Phú Quý

#### Những khó khăn gặp phải

**Khó khăn kĩ thuật**: 
- Chưa thành thạo với kĩ thuật groupby, aggregate trong thư viện pandas
- Không quen với việc trực quan hóa bằng Matplotlib và Seaborn, cần phải thử nhiều màu sắc và kích thước.

**Khó khăn phân tích**: 
- **Câu 4:** Các bài hát chết nhiều (popularity = 0), khó phân tích cho câu hỏi so sánh giữa 2 nhóm Classic songs và Recent Songs
- **Câu 5:** Chiến lược sản xuất âm nhạc của nghệ sĩ phần lớn (~80% dataset) là Low (<50 bài), trong khi lại có những ca sĩ có rất nhiều bài hát, khiến cho việc chọn các bin cho mỗi chiến lược gặp khó khăn.




#### Học hỏi và phát triển
**Kỹ năng xử lý dữ liệu:** Thành thạo hơn với groupby, aggregate trong pandas và hiểu cách chọn phép thống kê phù hợp theo mục tiêu phân tích.

**Kỹ năng ML**: Học được rằng XGBoost và LightGBM là tốt và phù hợp với dữ liệu bảng. 

**Trực quan hóa dữ liệu:** Biết cách sử dụng Matplotlib/Seaborn hiệu quả hơn, chọn biểu đồ, màu sắc và kích thước phù hợp để truyền tải insight rõ ràng.

**Tư duy phân tích:** Học được cách xử lý phân phối dữ liệu mất cân bằng, điều chỉnh chiến lược phân nhóm và rút ra kết luận dựa trên dữ liệu thay vì giả định.

**Làm việc nhóm:** Rèn luyện khả năng trao đổi, phân chia công việc và hỗ trợ lẫn nhau.

**Nhận thức về dữ liệu âm nhạc:** Bất ngờ khi dữ liệu âm nhạc cho thấy nhiều insight phức tạp hơn tưởng tượng, đặc biệt là ảnh hưởng chiến lược sản xuất nhạc rất có ảnh hưởng đến sự nổi tiếng và fanbase

### C. Trần Chí Nguyên

#### Những khó khăn gặp phải

**Khó khăn kỹ thuật:**
- Gặp khó trong việc tinh chỉnh biểu đồ với matplotlib và seaborn vì có quá nhiều tùy chọn cho màu sắc, kích thước, và cách bố trí. Phải thử nghiệm nhiều lần mới có biểu đồ đẹp và dễ hiểu.
- Gặp vấn đề khi gom nhóm dữ liệu để phân tích (groupby, aggregate) đúng cách và hiệu quả.

**Khó khăn phân tích:**
- **Câu 1 (So sánh loại album):** Ban đầu chỉ so sánh giá trị trung bình (mean), nhưng sau đó nhận ra cần phải xem thêm median, distribution, và biểu đồ boxplot để hiểu đầy đủ về outliers và cách phân bố dữ liệu. Ngoài ra còn cần xem xu hướng qua các năm để thấy rõ sự thay đổi.
- **Câu 2 (Explicit content):** Ban đầu giả định rằng nội dung nhạy cảm sẽ làm giảm popularity vì có thể bị hạn chế, nhưng kết quả phân tích cho thấy ngược lại hoàn toàn - các bài hát explicit lại phổ biến hơn.

**Quản lý thời gian:**
- Mất nhiều thời gian hơn dự kiến cho việc vẽ và làm đẹp các biểu đồ visualization. Nên dành thời gian chuẩn bị cho công đoạn này ngay từ đầu.
- Học được bài học quan trọng: phải ưu tiên làm phần tiền xử lý dữ liệu (data preprocessing) trước, sau đó mới phân tích các câu hỏi. Nếu không làm tốt bước này, dữ liệu sẽ lộn xộn và phải làm lại nhiều lần.

#### Học hỏi và phát triển

**Kỹ năng kỹ thuật:**
- Học được cách sử dụng matplotlib và seaborn để tạo các loại biểu đồ phức tạp: bar chart (biểu đồ cột), boxplot (biểu đồ hộp), violin plot, và line chart (biểu đồ đường).
- Hiểu rõ và sâu hơn về các hàm của pandas như `groupby()`, `agg()`, và `value_counts()` để tính toán thống kê mô tả một cách nhanh chóng và hiệu quả.
- Biết cách xử lý và chuyển đổi dữ liệu datetime để trích xuất thông tin về năm và thời gian.

**Tư duy phân tích:**
- Học được rằng không nên chỉ nhìn vào giá trị trung bình (mean) mà còn phải xem thêm median, độ lệch chuẩn (std), và phân bố (distribution) để hiểu đầy đủ về dữ liệu. Ví dụ: mean có thể bị ảnh hưởng nhiều bởi outliers.
- Rèn luyện khả năng nhìn nhận và rút ra insights từ các biểu đồ một cách trực quan.
- Nhận ra tầm quan trọng của visualization - một biểu đồ tốt có thể truyền đạt thông tin nhanh hơn và hiệu quả hơn nhiều so với bảng số liệu.
- Hiểu rằng kết quả phân tích có thể khác với giả định ban đầu, và đó là điều hoàn toàn bình thường trong nghiên cứu thực tế. Điều quan trọng là phải giải thích được tại sao lại có sự khác biệt đó.

**Khoảng cách giữa lý thuyết và thực tế:**
- Trong lý thuyết được học về các loại biểu đồ và khi nào nên dùng, nhưng trong thực tế phải tự quyết định biểu đồ nào phù hợp nhất cho từng câu hỏi cụ thể. Không có công thức cố định hay câu trả lời đúng tuyệt đối.
- Dữ liệu thực tế luôn có các vấn đề như missing values (giá trị thiếu), outliers (giá trị ngoại lai), và nhiều vấn đề khác. Cần phải linh hoạt và sáng tạo trong cách xử lý, không thể áp dụng máy móc theo sách vở.

**Những điều bất ngờ nhất:**
- Bất ngờ nhất là các bài hát có nội dung nhạy cảm (explicit content) lại có độ phổ biến cao hơn so với các bài hát thông thường. Ban đầu nghĩ rằng chúng sẽ bị hạn chế và ít phổ biến hơn, nhưng thực tế lại ngược lại.
- Album phát hành đầy đủ lại phổ biến hơn Single, mặc dù hiện nay xu hướng phát hành Single đang rất thịnh hành. Điều này cho thấy chất lượng và sự đầu tư vào một sản phẩm hoàn chỉnh vẫn quan trọng hơn việc chỉ tập trung vào số lượng.

**Khó khăn kỹ thuật:** Khó khăn trong việc chọn đặc trưng hợp lý từ dữ liệu có sẵn. Dataset thiếu nhiều thông tin quan trọng (đặc trưng âm thanh, marketing, vị trí trong danh sách phát) nên $R^2$ chỉ đạt ~0.36.

**Khó khăn phân tích:**
- **Tìm độ dài vàng**: Phải thử nghiệm nhiều cách phân khoảng khác nhau (1 phút, 0.5 phút, 0.25 phút) để tìm khoảng thời lượng vàng chính xác. Cân nhắc giữa độ chi tiết và kích thước mẫu.
- **Giải thích độ quan trọng đặc trưng**: Khó giải thích tại sao `artist_popularity` quan trọng hơn `track_duration_min`, phải phân tích sâu hơn về mối quan hệ giữa các biến.
- **Lựa chọn mô hình**: So sánh nhiều mô hình (Hồi quy tuyến tính, Random Forest, Gradient Boosting, Extra Trees) để tìm mô hình tốt nhất cho từng phân khúc độ phổ biến.

**Giải quyết:**
- Tham khảo ví dụ về độ quan trọng đặc trưng và cách giải thích mô hình
- Làm việc nhóm để brainstorm các đặc trưng có thể tạo ra từ dữ liệu hiện có

#### Học hỏi và phát triển

**Kỹ năng kỹ thuật:**
- **Pandas & NumPy**: Thành thạo hơn trong xử lý dữ liệu, nhóm dữ liệu, phân khoảng, gộp tập dữ liệu
- **Scikit-learn**: Hiểu sâu về quy trình máy học, chia dữ liệu, xác thực chéo, chuẩn hóa, các mô hình hồi quy
- **Git & cộng tác**: Làm quen với commit, push, xử lý xung đột, quy trình làm việc nhóm

**Hiểu biết về Máy học:**
- **Xác thực chéo**: Hiểu tầm quan trọng của xác thực chéo K-Fold trong việc đánh giá độ tin cậy mô hình, tránh quá khớp
- **Rò rỉ dữ liệu**: Học được khái niệm nghiêm trọng của rò rỉ dữ liệu và cách phòng tránh
- **Độ quan trọng đặc trưng**: Hiểu cách các mô hình cây quyết định tính toán độ quan trọng đặc trưng và ý nghĩa thực tế
- **Giải thích mô hình**: $R^2$ cao không phải lúc nào cũng tốt, phải xem xét bối cảnh và các đặc trưng còn thiếu

### C. Trần Chí Nguyên

#### Những khó khăn gặp phải

**Khó khăn kỹ thuật:**
- Gặp khó trong việc tinh chỉnh biểu đồ với matplotlib và seaborn vì có quá nhiều tùy chọn cho màu sắc, kích thước, và cách bố trí. Phải thử nghiệm nhiều lần mới có biểu đồ đẹp và dễ hiểu.
- Gặp vấn đề khi gom nhóm dữ liệu để phân tích (groupby, aggregate) đúng cách và hiệu quả.

**Khó khăn phân tích:**
- **Câu 1 (So sánh loại album):** Ban đầu chỉ so sánh giá trị trung bình (mean), nhưng sau đó nhận ra cần phải xem thêm median, distribution, và biểu đồ boxplot để hiểu đầy đủ về outliers và cách phân bố dữ liệu. Ngoài ra còn cần xem xu hướng qua các năm để thấy rõ sự thay đổi.
- **Câu 2 (Explicit content):** Ban đầu giả định rằng nội dung nhạy cảm sẽ làm giảm popularity vì có thể bị hạn chế, nhưng kết quả phân tích cho thấy ngược lại hoàn toàn - các bài hát explicit lại phổ biến hơn.

**Quản lý thời gian:**
- Mất nhiều thời gian hơn dự kiến cho việc vẽ và làm đẹp các biểu đồ visualization. Nên dành thời gian chuẩn bị cho công đoạn này ngay từ đầu.
- Học được bài học quan trọng: phải ưu tiên làm phần tiền xử lý dữ liệu (data preprocessing) trước, sau đó mới phân tích các câu hỏi. Nếu không làm tốt bước này, dữ liệu sẽ lộn xộn và phải làm lại nhiều lần.

#### Học hỏi và phát triển

**Kỹ năng kỹ thuật:**
- Học được cách sử dụng matplotlib và seaborn để tạo các loại biểu đồ phức tạp: bar chart (biểu đồ cột), boxplot (biểu đồ hộp), violin plot, và line chart (biểu đồ đường).
- Hiểu rõ và sâu hơn về các hàm của pandas như `groupby()`, `agg()`, và `value_counts()` để tính toán thống kê mô tả một cách nhanh chóng và hiệu quả.
- Biết cách xử lý và chuyển đổi dữ liệu datetime để trích xuất thông tin về năm và thời gian.

**Tư duy phân tích:**
- Học được rằng không nên chỉ nhìn vào giá trị trung bình (mean) mà còn phải xem thêm median, độ lệch chuẩn (std), và phân bố (distribution) để hiểu đầy đủ về dữ liệu. Ví dụ: mean có thể bị ảnh hưởng nhiều bởi outliers.
- Rèn luyện khả năng nhìn nhận và rút ra insights từ các biểu đồ một cách trực quan.
- Nhận ra tầm quan trọng của visualization - một biểu đồ tốt có thể truyền đạt thông tin nhanh hơn và hiệu quả hơn nhiều so với bảng số liệu.
- Hiểu rằng kết quả phân tích có thể khác với giả định ban đầu, và đó là điều hoàn toàn bình thường trong nghiên cứu thực tế. Điều quan trọng là phải giải thích được tại sao lại có sự khác biệt đó.

**Khoảng cách giữa lý thuyết và thực tế:**
- Trong lý thuyết được học về các loại biểu đồ và khi nào nên dùng, nhưng trong thực tế phải tự quyết định biểu đồ nào phù hợp nhất cho từng câu hỏi cụ thể. Không có công thức cố định hay câu trả lời đúng tuyệt đối.
- Dữ liệu thực tế luôn có các vấn đề như missing values (giá trị thiếu), outliers (giá trị ngoại lai), và nhiều vấn đề khác. Cần phải linh hoạt và sáng tạo trong cách xử lý, không thể áp dụng máy móc theo sách vở.

**Những điều bất ngờ nhất:**
- Bất ngờ nhất là các bài hát có nội dung nhạy cảm (explicit content) lại có độ phổ biến cao hơn so với các bài hát thông thường. Ban đầu nghĩ rằng chúng sẽ bị hạn chế và ít phổ biến hơn, nhưng thực tế lại ngược lại.
- Album phát hành đầy đủ lại phổ biến hơn Single, mặc dù hiện nay xu hướng phát hành Single đang rất thịnh hành. Điều này cho thấy chất lượng và sự đầu tư vào một sản phẩm hoàn chỉnh vẫn quan trọng hơn việc chỉ tập trung vào số lượng.
