# III. Đặt câu hỏi

Phần này thực hiện đề xuất các câu hỏi có ý nghĩa trong việc phân tích dữ liệu.

## Câu hỏi 1: So sánh loại album

### Question

Loại album nào (`album_type`: album, compilation, single) có độ phổ biến cao hơn?

### Motivation & Benefits

**Tại sao câu hỏi này quan trọng:** Quyết định release strategy (single, album, compilation) ảnh hưởng trực tiếp đến độ phổ biến và doanh thu của bài hát trong thời đại streaming.

**Lợi ích của việc trả lời câu hỏi:** Giúp nghệ sĩ, record labels và marketing teams đưa ra chiến lược phát hành tối ưu và tối ưu hóa nguồn lực promotion dựa trên data thực tế.

**Ai quan tâm đến câu trả lời:** Nghệ sĩ mới lựa chọn release strategy, record labels planning schedules, marketing teams xây dựng campaigns, và A&R teams tư vấn cho nghệ sĩ.

**Vấn đề thực tế:** Trong streaming era với xu hướng phát hành single thay vì full albums, việc hiểu format nào mang lại popularity cao hơn giúp optimize release strategies.

## Câu hỏi 2: Explicit content và popularity

### Question

Những bài hát có explicit content (`explicit` = True) có độ phổ biến cao hơn hay thấp hơn so với những bài không explicit?

### Motivation & Benefits

**Tại sao câu hỏi này quan trọng:** Explicit content là controversial topic với debate về trade-off giữa artistic freedom và commercial reach, cần evidence-based insights thay vì chỉ assumptions.

**Lợi ích của việc trả lời câu hỏi:** Giúp nghệ sĩ và songwriters đưa ra quyết định về lyrical content, đồng thời cung cấp data cho labels xây dựng content guidelines và streaming platforms quản lý filtering.

**Ai quan tâm đến câu trả lời:** Nghệ sĩ cân nhắc content của lyrics, record labels xây dựng policies, streaming platforms managing filtering, và parents/educators quan tâm youth exposure.

**Vấn đề thực tế:** Nhiều nghệ sĩ struggle giữa maintain artistic integrity và maximize audience reach, cần hiểu actual impact của explicit content thay vì dựa vào intuition.

## Câu hỏi 3: Golden duration analysis

### Question

Có tồn tại một khoảng thời lượng "golden duration" nào cho track mà trong đó những bài hát có độ phổ biến trung bình cao nhất không? Nếu có, khoảng này là bao nhiêu phút?

### Motivation & Benefits

**Tại sao câu hỏi này quan trọng:** Track duration là yếu tố controllable trong production, và trong streaming era với short attention spans, tìm duration sweet spot có thể tác động lớn đến track success.

**Lợi ích của việc trả lời câu hỏi:** Producers và songwriters có data-driven guidelines về track length để tối ưu completion rate, balance artistic vision với audience preferences, và tăng cơ hội được add vào playlists.

**Ai quan tâm đến câu trả lời:** Music producers, audio engineers, songwriters planning structure, playlist curators, và streaming platform algorithm designers.

**Vấn đề thực tế:** Songs ngày càng ngắn để phù hợp streaming metrics và TikTok culture, nhưng liệu shorter songs có thực sự popular hơn hay có optimal range nào đó?

## Câu hỏi 4: Số lượng bài hát và độ nổi tiếng nghệ sĩ

### Question

Nghệ sĩ phát hành nhiều bài hát hơn thì có đạt được độ nổi tiếng trung bình cao hơn so với nghệ sĩ phát hành ít bài hơn không?

### Motivation & Benefits

**Tại sao câu hỏi này quan trọng:** Có debate về quality vs quantity approach trong music industry, việc hiểu mối quan hệ giữa release frequency và artist popularity giúp shape career strategies hiệu quả.

**Lợi ích của việc trả lời câu hỏi:** Nghệ sĩ plan release schedules tối ưu, labels allocate resources hiệu quả, và hiểu trade-off giữa consistency và perfection cho long-term career development.

**Ai quan tâm đến câu trả lời:** Independent artists building catalogs, artist managers planning trajectories, record labels với limited budgets, và music industry analysts.

**Vấn đề thực tế:** Streaming algorithms favor consistent releases, nhưng flooding market có thể dilute brand value—finding right balance giữa quantity và quality là critical cho sustainable success.

## Câu hỏi 5: Classic songs vs Recent songs

### Question

Những bài hát phát hành từ 2009-2015 (classic songs) có còn giữ được độ phổ biến cao so với những bài hát phát hành từ 2020-2025 (recent songs) không?

### Motivation & Benefits

**Tại sao câu hỏi này quan trọng:** Hiểu liệu older tracks maintain popularity hay bị newer releases overshadow là crucial cho đánh giá long-term value của music catalogs và reflect cultural shifts trong listening behavior.

**Lợi ích của việc trả lời câu hỏi:** Rights holders và publishers đánh giá catalog value, predict long-term revenue potential, và guide investment decisions trong music rights dựa trên consumption trends.

**Ai quan tâm đến câu trả lời:** Music catalog investors, streaming platforms optimizing recommendations, nostalgic content creators, music historians, và artists với established back catalogs.

**Vấn đề thực tế:** Với daily explosion của new releases, có concern về older music getting lost, nhưng TikTok trends đã brought old songs back—understanding dynamics này helps catalog management và reissue strategies.

## Câu hỏi 6: Đặc trưng quan trọng theo từng phân khúc độ nổi tiếng

### Question

Những đặc trưng nào (features) quan trọng nhất thay đổi theo từng cụm (segment) độ nổi tiếng? Liệu mỗi phân khúc popularity có các yếu tố quyết định riêng biệt không?

### Motivation & Benefits

**Tại sao câu hỏi này quan trọng:** Không phải tracks đều target cùng popularity level, việc hiểu factors drive success ở mỗi tier (niche, mid-tier, mainstream) giúp tailor strategies appropriately thay vì one-size-fits-all approach.

**Lợi ích của việc trả lời câu hỏi:** Personalized strategies cho different career stages, understand market segmentation, optimize resources based on realistic goals, và identify pathways từ niche đến mainstream.

**Ai quan tâm đến câu trả lời:** Strategic planners trong labels, artists ở different stages, marketing teams với targeted campaigns, data scientists building recommendations, và investors analyzing opportunities.

**Vấn đề thực tế:** Emerging artists cần different strategies so với established superstars—genre, demographics, và platform preferences vary by popularity tier, cần segmentation analysis cho nuanced strategies.

## Câu hỏi 7: Dự đoán độ phổ biến bài hát (Machine Learning)

### Question

Liệu có thể dự đoán được độ nổi tiếng (popularity) của một bài hát dựa vào các thông tin có sẵn trong dataset này không? Nếu có thể, độ chính xác của mô hình dự đoán đạt đến mức nào và yếu tố nào có ảnh hưởng mạnh nhất?

### Motivation & Benefits

**Tại sao câu hỏi này quan trọng:** Khả năng predict success trước release là holy grail của music industry—với massive investments vào production và marketing, predictive models giúp minimize risks và reveal hidden patterns mà intuition có thể miss.

**Lợi ích của việc trả lời câu hỏi:** A&R teams screen demos hiệu quả, labels prioritize marketing budgets, producers optimize decisions, và feature importance analysis reveals actionable insights cho toàn industry.

**Ai quan tâm đến câu trả lời:** Label executives making signings, marketing departments allocating budgets, producers/songwriters trong creative process, investors trong music rights, streaming platforms, và music tech startups.

**Vấn đề thực tế:** Industry traditionally relies on subjective judgment—data-driven approaches complement human expertise, de-risk investments, identify overlooked talent, và even imperfect predictions provide directional guidance.
